# MiniCLIAgent 工程化链路 Demo

这份 notebook 用来演示 MiniCLIAgent 里更偏工程实践的三条能力：

- task
- worktree
- team

建议在 git 工作区中执行。

## 0. 开始前

请先确认：

- 当前目录是仓库根目录
- `.env` 已配置
- 当前 `MINICLIAGENT_WORKSPACE` 指向一个 git repo
- 默认测试已经通过

```python
!python -m minicliagent.cli.main tasks create --subject "demo-task" --description "learn engineering flow"
```

这一步先创建一个任务，后面会把 worktree 绑定到它。

```python
!python -m minicliagent.cli.main tasks list
```

```python
!python -m minicliagent.cli.main worktree list
```

如果当前还没有 MiniCLIAgent 管理过的 worktree，列表可能是空的。

```python
!python -m minicliagent.cli.main worktree create --name demo-wt --branch wt/demo-wt --task-id 1
```

这一步创建一个 worktree，并把它绑定到 task 1。

```python
!python -m minicliagent.cli.main worktree list
```

```python
!python -m minicliagent.cli.main tasks update --task-id 1 --status in_progress --owner alice
```

现在任务已经进入进行中状态，并被分配给 `alice`。

```python
!python -m minicliagent.cli.main team send --from lead --to alice --content "Please work on task 1 in demo-wt"
```

```python
!python -m minicliagent.cli.main team inbox --name alice
```

这两步模拟了一个很轻量的团队协作流：

- 任务已创建
- worktree 已绑定
- owner 已分配
- teammate 收到消息

现在你可以再去看这些状态目录：

- `.minicliagent/tasks/`
- `.minicliagent/worktrees/`
- `.minicliagent/team/`

这就是 MiniCLIAgent 比“只会单轮聊天”的 demo 更工程化的地方。
